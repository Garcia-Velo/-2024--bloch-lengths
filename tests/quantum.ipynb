```python
# Numerical and scientific python programming
import numpy as np

import matplotlib.pyplot as plt

# Local importations
from moments.quantum import generate_rand_dm, compute_is_valid_dm
```

# Density validation and matrix generation

```python
I = np.eye(4)

print("Invalid trace")
is_valid, info = compute_is_valid_dm(I)
print(info["trace_ok"])

I = np.eye(4)/np.trace(I)
A = I.copy().astype(complex)
A[1, 1] = 1j

print("Invalid Hermitian")
is_valid, info = compute_is_valid_dm(A)
print(info["hermitian_ok"])

A[1, 1] = -1

print("Invalid PSD")
is_valid, info = compute_is_valid_dm(A)
print(info["psd_ok"])

print("Valid density matrix")
is_valid, info = compute_is_valid_dm(I)
print(info["trace_ok"])
```

```python
all_valid = True
for i in range(int(1e4)):
    rho = generate_rand_dm(4, 4)
    is_valid = compute_is_valid_dm(rho)
    if not is_valid:
        all_valid = False

if all_valid:
    print("All generate density matrices are valid.")
else:
    print("Some generated density matrices are invalid.")
```
