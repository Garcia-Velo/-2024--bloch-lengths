# Test the code from file `quantum.py`

This means testing the function `compute_is_valid_dm`. It checks whether `rho` is a valid densit matrix or not. It takes two inputs
- rho: np.ndarray. A square matrix.
- tol: float = 1e-10. Tolerance for numerical comparisons.
It returns a tuple with two elements:
- float. `True` if `rho` is a valid density matrix, `False` if not.
- dict[str, Any]. A dictionary containing the diagnostic invormation. This includes:
    - "trace": The numerical value of the trace of `rho´.
    - "trace_ok": `True` if the trace of `rho` is $1$ under the numerical tolerance, `False` if not.
    - "hermitian_ok": `True` if the `rho` is Hermitian under the numerical tolerance, `False` if not.
    - "psd_ok": `True` if the `rho` is positive semi-definite under the numerical tolerance, `False` if not.
    - "min_eig": Value of the minimum eigenvalue. Only if the matrix is Hermitian, `None` otherwise.

# Importations

```python
# Numerical and scientific python programming
import numpy as np

# Local importations
from moments.quantum import compute_is_valid_dm
```

# Density validation and matrix generation

To test it we just design some simple scenarios where the matrices are known to fail to meet the conditions.

```python
I = np.eye(4)

print("- Invalid trace")
is_valid, info = compute_is_valid_dm(I)
print(info["trace_ok"])

I = np.eye(4)/np.trace(I)
A = I.copy().astype(complex)
A[1, 1] = 1j

print("\n- Invalid Hermitian")
is_valid, info = compute_is_valid_dm(A)
print(info["hermitian_ok"])

A[1, 1] = -1

print("\n- Invalid PSD")
is_valid, info = compute_is_valid_dm(A)
print(info["psd_ok"])

print("\n- Valid density matrix")
is_valid, info = compute_is_valid_dm(I)
print(info["trace_ok"])
```
