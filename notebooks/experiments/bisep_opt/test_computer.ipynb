# Importations

```python
import numpy as np
from bisep_computer import biseparable_state
```

# Definitions

```python
phi_0 = np.array([1, 0]).reshape(2, 1)
phi_1 = np.array([0, 1]).reshape(2, 1)

phi_00 = np.kron(phi_0, phi_0)
phi_11 = np.kron(phi_1, phi_1)
```

```python
p = np.array([1/3]*3)
q_A = q_B = q_C= np.array([0.5]*2)
```

```python
psiA_1 = psiB_1 = 0.5 * (- np.sqrt(3) * phi_0 + phi_1)
psiA_2 = psiB_2 = 0.5 * (np.sqrt(3) * phi_0 + phi_1)

psiBC_1 = psiAC_1 = 0.5 * (np.sqrt(3) * phi_00 + phi_11)
psiBC_2 = psiAC_2 = 0.5 * (- np.sqrt(3) * phi_00 + phi_11)

psiC_1 = 0.5 * (- np.sqrt(3) * 1j * phi_0 + phi_1)
psiC_2 = 0.5 * (np.sqrt(3) * 1j * phi_0 + phi_1)

psiAB_1 = 0.5 * (- np.sqrt(3) * 1j * phi_00 + phi_11)
psiAB_2 = 0.5 * (np.sqrt(3) * 1j * phi_00 + phi_11)

PSI_A = PSI_B = [psiA_1, psiA_2]
PSI_C = [psiC_1, psiC_2]

PSI_BC = PSI_AC = [psiAC_1, psiAC_2]
PSI_AB = [psiAB_1, psiAB_2]
```

```python
print("Subsystem A:")
print("\n- A:")
for psi_a in PSI_A:
    print(np.round(psi_a, 2))

print("\n- BC:")
for psi_bc in PSI_BC:
    print(np.round(psi_bc, 2))

print("\nSubsystem B:")
print("\n- B:")
for psi_b in PSI_B:
    print(np.round(psi_b, 2))

print("\n- AC:")
for psi_ac in PSI_AC:
    print(np.round(psi_ac, 2))

print("\nSubsystem C:")
print("\n- C:")
for psi_c in PSI_C:
    print(np.round(psi_c, 2))

print("\n- BC:")
for psi_ab in PSI_AB:
    print(np.round(psi_ab, 2))
```

# Compute state

```python
rho = biseparable_state(p,
                        PSI_A, q_A, PSI_BC,
                        PSI_B, q_B, PSI_AC,
                        PSI_C, q_C, PSI_AB)
```

```python
print(rho.real)
```
