# Tests

```python
N = 2
dim = [2]*N

psi = generate_GHZ_state(N)
rho = compute_outer_product(psi, psi)

result = find_decomposable_witness(dim, rho)
print(result.minval)
```

# Noise thresholds

```python
def compute_noise_state(rho, p):
    return (1 - p) * rho + p * (1/d) * np.eye(d)

def test_noise_state_ent(rho, p_lim):
    wit_result = []
    p_vec = np.linspace(p_lim[0], p_lim[1], 100)
    for p in p_vec:
        rho_n = compute_noise_state(rho, p)
        result = find_decomposable_witness(dim, rho_n)
        wit_result.append(result.minval)
        pdx = np.abs(wit_result).argmin()
    return p_vec[pdx]
```

```python
N = 3
dim = [2]*N
d = int(np.prod(dim))

psi_GHZ = generate_GHZ_state(N)
rho_GHZ = compute_outer_product(psi_GHZ, psi_GHZ)

p_t = test_noise_state_ent(rho_GHZ, (0.569, 0.573))
print("Noise threshold: pt =", p_t)

psi_W = generate_W_state(N)
rho_W = compute_outer_product(psi_W, psi_W)

p_t = test_noise_state_ent(rho_W, (0.5, 0.54))
print("Noise threshold: pt =", p_t)
```
