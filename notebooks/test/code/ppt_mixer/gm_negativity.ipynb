# Importations

```python
# Numerical and scientific python programming
import numpy as np

# Local importations
from moments.quantum import compute_outer_product, generate_GHZ_state, generate_W_state
from moments.ppt_mixer.decomposable_witness import compute_gm_negativity
```

# Tests

```python
N = 4
dim = [2]*N
d = int(np.prod(dim))

print("W and GHZ states:\n")

psi_GHZ = generate_GHZ_state(N)
rho_GHZ = compute_outer_product(psi_GHZ, psi_GHZ)

N_GHZ = compute_gm_negativity(dim, rho_GHZ)
print(f"N_GHZ = {N_GHZ:.4f}")

psi_W = generate_W_state(N)
rho_W = compute_outer_product(psi_W, psi_W)

N_W = compute_gm_negativity(dim, rho_W)
print(f"N_W = {N_W:.4f}")
```

```python
print("Some biseparable states:\n")

PSI_GHZ = [generate_GHZ_state(n) for n in range(1, N)]
PSI_W = [generate_W_state(n) for n in range(1, N)]

PSI_BI = [np.kron(PSI_GHZ[0], PSI_GHZ[2]),
np.kron(PSI_GHZ[1], PSI_GHZ[1]),
np.kron(PSI_W[0], PSI_W[2]),
np.kron(PSI_W[1], PSI_W[1]),
np.kron(PSI_GHZ[0], PSI_W[2]),
np.kron(PSI_GHZ[2], PSI_W[0]),
np.kron(PSI_GHZ[1], PSI_W[1])
]

RHO_BI = [compute_outer_product(psi_bi, psi_bi) for psi_bi in PSI_BI]

N_BI = [compute_gm_negativity(dim, rho_bi) for rho_bi in RHO_BI]

for n_bi in N_BI:
    print(f"N = {np.abs(n_bi):.4f}")
```
